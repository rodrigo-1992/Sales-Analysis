```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
NOTEBOOK_DIR = Path(os.getcwd())
PROJECT_ROOT = NOTEBOOK_DIR.parent
RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
PROCESSED_DIR = PROJECT_ROOT / 'data' / 'processed'
```

```python
def carregar_dados_csv(nome_arquivo):

    caminho = RAW_DIR / nome_arquivo
    if caminho.exists():
        try:
            df = pd.read_csv(caminho)
            return df
        except Exception as e:
            print(f"Erro ao carregar o arquivo {nome_arquivo}: {e}")
            return None
    else:
        print(f"Arquivo {nome_arquivo} não encontrado no diretório {RAW_DIR}.")
        return None
    
```

```python
def relatorio_limpeza(df, nome_tabela):

    print(f"Relatorio de Limpeza de Dados para a tabela: {nome_tabela.upper()}")
    print(f"{'='*60}")
    print(f"Shape original: {df.shape}")
    print(f"Memoria utilizada: {df.memory_usage(deep=True).sum() / 1024 ** 2:.2f} MB")
    print(f"\n Valores Nulos:")
    nulos = df.isnull().sum()
    if nulos.sum() > 0:
        print(nulos[nulos > 0])
    else:
        print("Nenhum valor nulo encontrado.")
    print(f"Duplicatas: {df.duplicated().sum()}")
    print(f"Tipos de dados:")
    print(df.dtypes)
    return nulos
```

```python
def salvar_dataset(df, nome_arquivo):
    caminho = PROCESSED_DIR / nome_arquivo
    df.to_csv(caminho, index=False)
    print(f"Dataset salvo em: {caminho}")
    return caminho
```

```python
arquivos = {
    'orders': 'olist_orders_dataset.csv',
    'customers': 'olist_customers_dataset.csv',
    'products': 'olist_products_dataset.csv',
    'order_items': 'olist_order_items_dataset.csv',
    'order_payments': 'olist_order_payments_dataset.csv',
    'order_reviews': 'olist_order_reviews_dataset.csv',
    'sellers': 'olist_sellers_dataset.csv'
}

dados = {}
for nome, arquivo in arquivos.items():
    df = carregar_dados_csv(arquivo)
    if df is not None:
        dados[nome] = df
```

```python
print("Limpando dados da tabela Orders")
print("="*60)

orders = dados['orders'].copy()

# Converter datas
colunas_data = ['order_purchase_timestamp', 
                'order_approved_at', 
                'order_delivered_carrier_date', 
                'order_delivered_customer_date', 
                'order_estimated_delivery_date']

for col in colunas_data:
    if col in orders.columns:
        orders[col] = pd.to_datetime(orders[col], errors='coerce')
        print(f"Coluna {col} convertida para datetime.")

# Tratando valores nulos

orders['order_approved_at'] = orders['order_approved_at'].fillna(orders['order_purchase_timestamp'])
orders['order_delivered_carrier_date'] = orders['order_delivered_carrier_date'].fillna(orders['order_estimated_delivery_date'])
orders['order_delivered_customer_date'] = orders['order_delivered_customer_date'].fillna(orders['order_estimated_delivery_date'])

# Remove pedidos com status indefinido

orders = orders[orders['order_status'].notna()]

# Remover duplicatas

orders = orders.drop_duplicates(subset=['order_id'])

# Criar colunas derivadas

orders['order_purchase_date'] = orders['order_purchase_timestamp'].dt.date
orders['order_purchase_year'] = orders['order_purchase_timestamp'].dt.year
orders['order_purchase_month'] = orders['order_purchase_timestamp'].dt.month
orders['order_purchase_day'] = orders['order_purchase_timestamp'].dt.day
orders['order_purchase_weekday'] = orders['order_purchase_timestamp'].dt.day_name()
orders['order_purchase_hour'] = orders['order_purchase_timestamp'].dt.hour

# Calcular tempo de entrega em dias

if 'order_delivered_customer_date' in orders.columns and 'order_purchase_timestamp' in orders.columns:
    orders['delivery_time_days'] = (
        orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']
        ).dt.days

# Salvar dataset limpo

salvar_dataset(orders, 'orders_clean.csv')
dados['orders'] = orders
```

```python
# Limpeza de dados da tabela Customers

customers = dados['customers'].copy()

# Remover duplicatas

customers = customers.drop_duplicates(subset=['customer_id'])

# Padronização de texto

for col in ['customer_city', 'customer_state']:
    if col in customers.columns:
        customers[col] = customers[col].str.upper().str.strip()

# Salvar dataset

salvar_dataset(customers,'customers_clean.csv')
dados['customers'] = customers
```

```python
# Limpeza de dados da tabela Products

products = dados['products'].copy()

# Preenchendo valores nulos com 'unknown'

products['product_category_name'] = products['product_category_name'].fillna('unknown')

# Preencher dimensões nulas com mediana
for col in ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']:
    if col in products.columns:
        products[col] = products[col].fillna(products[col].median())

# Padronizar texto
products['product_category_name'] = products['product_category_name'].str.lower().str.strip()

# Remover duplicatas
products = products.drop_duplicates(subset=['product_id'])

# Remover produtos sem id

products = products[products['product_id'].notna()]

salvar_dataset(products,'products_clean.csv')
```

```python
# Limpeza de dados da Tabela Orders_item

order_items = dados['order_items'].copy()

# Tratar valores nulos
order_items['price'] = order_items['price'].fillna(order_items['price'].median())
order_items['freight_value'] = order_items['freight_value'].fillna(order_items['freight_value'].median())

# Remover linhas sem order_id ou product_id
order_items = order_items[order_items['order_id'].notna()]
order_items = order_items[order_items['product_id'].notna()]
order_items = order_items[order_items['seller_id'].notna()]

# Remover duplicatas
order_items = order_items.drop_duplicates()

# Criando colunas derivadas
order_items['total_value'] = order_items['price'] + order_items['freight_value']

salvar_dataset(order_items, 'order_items_clean.csv')
dados['order_items'] = order_items
```

```python
# Limpeza da Tabela Order_payments

order_payments = dados['order_payments'].copy()

# Tratar valores nulos

order_payments['payment_type'] = order_payments['payment_type'].fillna('unknown')
order_payments['payment_installments'] = order_payments['payment_installments'].fillna(1)
order_payments['payment_value'] = order_payments['payment_value'].fillna(order_payments['payment_value'].median())

# Padronizar texto
order_payments['payment_type'] = order_payments['payment_type'].str.lower().str.strip()

# Remover duplicatas
order_payments = order_payments.drop_duplicates()

# Agrupar pagamentos do mesmo pedido
order_payments_grouped = order_payments.groupby('order_id').agg({
    'payment_type': lambda x: ','.join(x.unique()),
    'payment_installments': 'max',
    'payment_value': 'sum'
}).reset_index()

salvar_dataset(order_payments_grouped, 'order_payments_clean.csv')
dados['order_payments'] = order_payments_grouped
```

```python
# Limpeza Order_Review
order_reviews = dados['order_reviews'].copy()

if 'review_creation_date' in order_reviews.columns:
    order_reviews['review_creation_date'] = pd.to_datetime(
        order_reviews['review_creation_date'], errors='coerce'
    )
if 'review_answer_timestamp' in order_reviews.columns:
    order_reviews['review_answer_timestamp'] = pd.to_datetime(
        order_reviews['review_answer_timestamp'], errors='coerce'
    )

# Preencher valores nulos
order_reviews['review_score'] = order_reviews['review_score'].fillna(order_reviews['review_score'].median())
order_reviews['review_comment_title'] = order_reviews['review_comment_title'].fillna('')
order_reviews['review_comment_message'] = order_reviews['review_comment_message'].fillna('')

# Remover duplicatas
order_reviews = order_reviews.drop_duplicates(subset=['review_id'])

# Calculo do tempo de resposta
if 'review_creation_date' in order_reviews.columns and 'review_answer_timestamp' in order_reviews.columns:
    order_reviews['review_response_time_days'] = (
        order_reviews['review_answer_timestamp'] - order_reviews['review_creation_date']
    ).dt.days

salvar_dataset(order_reviews, 'order_reviews_clean.csv')
dados['order_reviews'] = order_reviews
```

```python
# Limpeza Tabela Sellers

sellers = dados['sellers'].copy()

# Remover duplicatas
sellers = sellers.drop_duplicates(subset=['seller_id'])

# Padronização de texto
for col in ['seller_city', 'seller_state']:
    if col in sellers.columns:
        sellers[col] = sellers[col].str.upper().str.strip()

salvar_dataset(sellers,'sellers_clean.csv')
dados['sellers'] = sellers
```
